# Get Balanced Data

```python
def get_reviews(app_id, app_name, score):
  # Get the reviews
  result, continuation_token = reviews(
      app_id,
      lang='en',
      country='us',
      sort=Sort.NEWEST,
      count=300,
      filter_score_with=score
  )

  # Get content and score only
  df = pd.DataFrame(result)
  df_selected = df[['content', 'score']].copy()

  return df_selected
```

```python
all_df_balanced = []
df_balanced = pd.DataFrame()

apps = {
    "Lichess": "org.lichess.mobileapp",
    "Chess.com": "com.chess"
}

for name, app_id in apps.items():
  for score in range(1, 6):
    all_df_balanced.append(get_reviews(app_id, name, score))

df_balanced = pd.concat(all_df_balanced, ignore_index=True)
```

```python
df_balanced.to_csv('balanced_data.csv')
```

# Get Imbalanced Data

```python
def get_reviews_2(app_id, app_name):
  # Get the reviews
  result, continuation_token = reviews(
      app_id,
      lang='en',
      country='us',
      sort=Sort.NEWEST,
      count=1500,
  )

  # Get content and score only
  df = pd.DataFrame(result)
  df_selected = df[['content', 'score']].copy()

  return df_selected
```
